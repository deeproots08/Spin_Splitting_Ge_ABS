```python
### Address for minimal model 1d python script ###

addr = Path.cwd().parent

### add the path to where 1d minimal model is stored to sys path###

sys.path.append(str(addr))

### import the py file ###

import Minimal_Model_1D as M

###defining constants##

hbar = 1.055*10**(-34);   #in J-s
ee = 1.602*10**(-19);     #in C
hbar_mev = hbar/ee*1000   # in meV-s
me = 9.11*10**(-31);      #in kg
hbm = hbar**2*1000/2/me/ee/(1e-18) #in meV (nm)**2

```

```python
NSM = 300          # in nm
NSC = 2000         # in nm
N = 2*NSC + NSM

MY = 1             # no of subbands
AX = 1             # lattice spacing in nm
LY = 100           # Width in nm
ALPHAT = 20        # meV-nm
DELTA = 1.65      # in meV
MEFF_SC = 0.023     # proximitized Ge 
MEFF_SM = 0.023     # Ge
OMEGA = 2*4*hbm/LY**2/MEFF_SC
PHI = 0.5*np.pi 


print('first subband energy =' , OMEGA , '\n', 'second subband energy =' , 2*OMEGA)

MU_SC =  0.8 + OMEGA   #in between two subbands
MU_SM =  0.8 + OMEGA   #in between two subbands
```

```python
D = M.oneD_Minimal_Model(N,NSM, NSC,MY,AX ,OMEGA, ALPHAT, DELTA, MEFF_SC, MEFF_SM, MU_SC, MU_SM,lambda1=1, lambda2=0)
H = D.H_sys(PHI, so='li')
E = eigsh(H,k=8,sigma=0, return_eigenvectors=False)
E2 = np.sort(E)
print(E2/DELTA)
```

```python
H_Normal = D.HN(so='li')
```

```python
nsc_data = np.full(N, DELTA*cmath.exp(1j*PHI))

DEL = diags(nsc_data)

DEL_MAT_SDOF = bmat([
           [None, DEL],
           [-DEL, None]]) 
```
