```python
import math, cmath
import numpy as np
import matplotlib.pyplot as plt
                                                                               
```

```python
def Harmonic_potential_iny(xp,yp,Vp):

    return Vp + yp**2
```

```python
def circular_pot(xp,yp):

    return xp**2 + yp**2
```

```python
def hyp_pot(xp,yp):

    return -np.sign(yp)*np.cosh(xp) + yp
```

```python
def gaussian(xp,yp):

    return np.exp(xp**2 + yp**2)
```

```python
x = np.linspace(-1,1,101)
y = np.linspace(-1,1,101)

X,Y = np.meshgrid(x,y)
Z = Harmonic_potential_iny(X,Y,1)
```

```python
print(X.shape, Y.shape, Z.shape)
```

```python
f = plt.figure(figsize=(10, 8))

ax = f.add_axes([0,0,1,1])

plt.contourf(X,Y,Z,101,cmap='GnBu_r')

ax.set_axis_off()
```
