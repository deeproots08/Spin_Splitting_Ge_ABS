```python
def compute_transforms_and_multiply(n, sigma, lambda_, a):
    gaussian_array = np.random.normal(loc=0, scale=sigma, size=n)
    ft_gaussian = np.fft.fft(gaussian_array)
    j = np.arange(n)
    exponential_array = sigma**2 * np.exp(-a * j / lambda_)
    ft_exponential = np.abs(np.fft.fft(exponential_array))
    product_of_coefficients = ft_gaussian * ft_exponential
    ift_result = np.fft.ifft(product_of_coefficients)
    variance = np.var(np.abs(ift_result))
    return gaussian_array, ft_gaussian, exponential_array, ft_exponential, ift_result
```

```python
# Parameters
n = 100  # Array size
sigma = 1.2  # Standard deviation
lambda_ = 1.0  # Decay constant
a = 0.1  # Lattice spacing
iterations = 10000


af, bf, cf, df, ef = compute_transforms_and_multiply(n, sigma, lambda_, a)

```
