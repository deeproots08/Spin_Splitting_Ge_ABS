### Code to Study variation of cubic spin orbit coupling constant calculated as per the Effective heavy hole model with z-confinement length

```python
Tot = 20

EP_eff_mass = np.empty((Tot,1))

EP_cubicSO_coeff = np.empty((Tot,1))

iter = 20

EF = 15  # in meV/nm

for p in range(20):

    PC = Eff_heavyhole_model_CSOC_parameters.Eff_hh_model_cubic_param(Lz = p+1)
    EP_eff_mass[p] = PC.Eff_mass(iter)
    EP_cubicSO_coeff[p] = PC.cubicSO_coeff(EF, iter)
    if p%10 == 0:

        print(p)
    
```

```python
class cubic_so2(Eff_hh_model_cubic_param):

    def cubicSO_coeff(self, Ef, pmax ):

        """
        Returns cubic spin orbit coupling factor
        
        Args:

            Ef (float) : Electric field in meV/nm
            pmax : maximum index till which you want to sum 
        """

        M = 0
        N = hbm**2*(3*self.g3)*(self.g2 + self.g3)*Ef*64/math.pi**2

        for p in np.arange(pmax)+1:

            p2 = 2*p
            A = 1/self.Del_E_phh(p2)/self.Del_E_olh() - 1/self.Del_E_phh(p2)/self.Del_E_plh(p2) -1/self.Del_E_olh()/self.Del_E_plh(p2)
            M = M + p2**2/(p2**2-1)**3*A

        return M*N
```
