```python
 import numpy as np
from scipy.sparse import csr_matrix
import random

# Define the size of the sparse matrix
n = 5  # Size of the matrix
nsm = 3  # Number of non-zero elements in each row

# Generate random values for the non-zero elements
values = [random.gauss(0, 1) for _ in range(n * nsm)]
values = np.array(values)

# Generate row indices for the non-zero elements
rows = np.repeat(np.arange(n), nsm)

# Generate column indices for the non-zero elements
cols = np.random.randint(0, n, size=n * nsm)

# Create the sparse matrix
sparse_matrix = csr_matrix((values, (rows, cols)), shape=(n, n))

print("Sparse Matrix:")
print(sparse_matrix.toarray())
```

```python
import numpy as np
import random
from scipy.sparse import diags

# Define parameters
m = 4  # Number of repetitions
nsc = 2  # Number of zero entries at the start and end
nsm = 3  # Number of non-zero entries in the middle

# Initialize the final array
final_array = np.array([])

# Repeat the process m times
for _ in range(m):
    # Generate random values for the non-zero entries
    values = [random.gauss(0, 1) for _ in range(nsm)]
    values = np.array(values)
    
    # Create the 1D array for this iteration
    array = np.concatenate((np.zeros(nsc), values, np.zeros(nsc)))
    
    # Concatenate the array to the final array
    final_array = np.concatenate((final_array, array))

# Create a sparse diagonal matrix from the final array
sparse_matrix = diags(final_array, 0, format='csr')

print("Sparse Diagonal Matrix:")
#print(sparse_matrix.toarray())
M =sparse_matrix.toarray()
```
