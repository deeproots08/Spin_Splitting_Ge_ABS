```python
nsc = 2
nsm = 5
n = nsm + 2*nsc
alpha = 20

m = 3

OP2d = O2d.Operators_2d(n, m, 1,1)

HA = mm2d.twoD_Minimal_Model(n,nsm,nsc,m,1,1, alphat=alpha).H_particle(so='cu')

H_kp = alpha*1j*(OP2d.KxCu()+3*1j*OP2d.KxSqKy() -3*OP2d.KxKySq()-1j*OP2d.KyCu())
H_km = alpha*(-1j)*(OP2d.KxCu()-3*1j*OP2d.KxSqKy() -3*OP2d.KxKySq()+1j*OP2d.KyCu())
H_k0 = hbm*(OP2d.KxSq()+OP2d.KySq())


# Construct a block matrix
HB = bmat([
        [H_k0, H_km],
        [H_kp, H_k0],
    ])
```

```python
nsc = 2
nsm = 5
n = nsm + 2*nsc
alpha = 20

m = 3

OP2d = O2d.Operators_2d(n, m, 1,1)

HA = mm2d.twoD_Minimal_Model(n,nsm,nsc,m,1,1, alphat=alpha).H_particle(so='li')

H_kp = alpha*(-1j*OP2d.Kx() + OP2d.Ky())
H_km = alpha*(1j*OP2d.Kx()+OP2d.Ky())
H_k0 = hbm*(OP2d.KxSq()+OP2d.KySq())


# Construct a block matrix
HB = bmat([
        [H_k0, H_km],
        [H_kp, H_k0],
    ])


d = HA != HB

print(d)

c = HA.toarray() == HB.toarray()
print(c.all())
```

```python
nsc = 2
nsm = 5
n = nsm + 2*nsc
alpha = 20

m = 3

OP2d = O2d.Operators_2d(n, m, 1,1)

HA = mm2d.twoD_Minimal_Model(n,nsm,nsc,m,1,1, alphat=alpha).H_sys(so='cu')

H_kp = alpha*1j*(OP2d.KxCu()+3*1j*OP2d.KxSqKy() -3*OP2d.KxKySq()-1j*OP2d.KyCu())
H_km = alpha*(-1j)*(OP2d.KxCu()-3*1j*OP2d.KxSqKy() -3*OP2d.KxKySq()+1j*OP2d.KyCu())
H_k0 = hbm*(OP2d.KxSq()+OP2d.KySq())


# Construct a block matrix
HP = bmat([
        [H_k0, H_km],
        [H_kp, H_k0],
    ])

HP = HP.tocsr()
print(HP.shape)
HB = bmat([
    [HP, None],
    [None, -HP.conjugate()]
])
print(HB.shape)
```

```python
d = HA != HB

print(d)

c = HA.toarray() == HB.toarray()
print(c.all())
```

```python
### comparing the spin splitting and first few energy eigenvalues of the minimal 2d model with the one d multiple subband model ###


import numpy as np 
from scipy.sparse import bmat, diags, csr_matrix
from scipy.sparse.linalg import eigsh,eigs
import Minimal_Model_2D as mm2d
import Minimal_Model_1D as mm1d
import Operators2D as O2d
```

```python
hbar = 1.055*10**(-34);   #in J-s
ee = 1.602*10**(-19);     #in C
hbar_mev = hbar/ee*1000   # in meV-s
me = 9.11*10**(-31);      #in kg
hbm = hbar**2*1000/2/me/ee/(1e-18) #in meV (nm)**2
```

```python
NSM = 300          # in nm
MY = 6             # no of subbands
AX = 1             # lattice spacing in nm
DELTA = 0.165      # in meV
#MEFF_SC = 0.05     # Ge 
#MEFF_SM = 0.05     # proximitized Ge
MEFF_SC = 0.023     # proximitized InAs
MEFF_SM = 0.023     # InAs 


NSC = 4000
N = 2*NSC + NSM
PHI = 0.5*np.pi 

#print('first subband energy =' , OMEGA , '\n', 'second subband energy =' , 2*OMEGA)

ALPHAT = 20                  # meV-nm # 20 for LSOC is within perturbative regime for widths atleast up to 100 nm 

LY = 50                      # in nm
OMEGA = 2*4*hbm/LY**2/MEFF_SC
print(OMEGA)

MU_SM = 0.8 + OMEGA/2        ## As per Andreev Aprroximation chemical pot >> superconducting gap 
MU_SC = 0.8 + OMEGA/2        ## constant potential over reference level across all widths

K = 12

#Ep_li = np.empty((LY.shape[0],K) )
#Kappa = np.empty((LY.shape[0],1) )

Fli = mm1d.oneD_Minimal_Model(N,NSM, NSC,MY,AX ,OMEGA, ALPHAT, DELTA, MEFF_SC, MEFF_SM, MU_SC, MU_SM)

Hli = Fli.H_sys(PHI, so = 'li')

print(Hli.shape)

Eli = eigsh(Hli ,k=K,sigma=0, return_eigenvectors=False)

Eli = np.sort(Eli)

Ep_Split_li = Eli[K//2+1] - Eli[K//2]

print(Ep_Split_li)
```
